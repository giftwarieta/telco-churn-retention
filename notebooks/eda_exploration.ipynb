```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual style
sns.set_theme(style="whitegrid", palette="muted")
%matplotlib inline

import warnings
warnings.filterwarnings('ignore')
```

```python
# Load and Initial Inspection

df = pd.read_csv('../data/WA_Fn-UseC_-Telco-Customer-Churn 2.csv')

print(f"Dataset Shape: {df.shape}")
df.head()
```

```python
df.describe(include='all')
```

```python
# Data Cleaning


# Handle TotalCharges (Convert string to float and fill missing values)
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
print(f"Missing values in TotalCharges: {df['TotalCharges'].isnull().sum()}")
df['TotalCharges'] = df['TotalCharges'].fillna(0)

# Convert Churn to binary for analysis
df['Churn_Numeric'] = df['Churn'].apply(lambda x: 1 if x == 'Yes' else 0)

# Drop CustomerID as it's not useful for EDA
df.drop('customerID', axis=1, inplace=True)
```

```python
# Target Variable Distribution

plt.figure(figsize=(6, 4))
sns.countplot(x='Churn', data=df)
plt.title('Distribution of Churn (Target Variable)')
plt.show()

churn_rate = df['Churn_Numeric'].mean()
print(f"Overall Churn Rate: {churn_rate:.2%}")
```

```python
# Insight

plt.figure(figsize=(10, 5))
sns.kdeplot(df[df['Churn'] == 'No']['tenure'], label='Loyal Customers', fill=True, color='blue')
sns.kdeplot(df[df['Churn'] == 'Yes']['tenure'], label='Churned Customers', fill=True, color='red')
plt.title('Tenure Distribution: When do we lose them?')
plt.xlabel('Months with Company')
plt.legend()
plt.show()

# Insight: High risk in months 1-6. Loyalty stabilizes after 24 months.
```

```python
# Contract & Payment Friction

fig, axes = plt.subplots(1, 2, figsize=(15, 6))

# Contract Type
sns.countplot(x='Contract', hue='Churn', data=df, ax=axes[0])
axes[0].set_title('Churn by Contract Type')

# Payment Method
sns.countplot(y='PaymentMethod', hue='Churn', data=df, ax=axes[1])
axes[1].set_title('Churn by Payment Method')

plt.tight_layout()
plt.show()

# Insight: Month-to-month contracts and Electronic Checks are the primary "leaks."
```

```python
# Pricing Gap

plt.figure(figsize=(10, 6))
sns.boxplot(x='InternetService', y='MonthlyCharges', hue='Churn', data=df)
plt.title('Monthly Charges vs Internet Service (Segmented by Churn)')
plt.show()

# Insight: Fiber Optic users pay more and churn more. 
# Are they leaving because of the price or service quality?
```

```python
# Correlation Analysis

# Select numeric columns and encoded target
numeric_cols = df.select_dtypes(include=[np.number])
corr_matrix = numeric_cols.corr()

plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Heatmap')
plt.show()
```

```python
# Feature Engineering

def tenure_bucket(tenure):
    if tenure <= 12: return '0-1 Year'
    elif tenure <= 24: return '1-2 Years'
    elif tenure <= 48: return '2-4 Years'
    else: return '4+ Years'

df['Tenure_Group'] = df['tenure'].apply(tenure_bucket)

plt.figure(figsize=(10, 5))
sns.barplot(x='Tenure_Group', y='Churn_Numeric', data=df, order=['0-1 Year', '1-2 Years', '2-4 Years', '4+ Years'])
plt.title('Churn Probability by Tenure Group')
plt.ylabel('Churn Rate')
plt.show()
```
